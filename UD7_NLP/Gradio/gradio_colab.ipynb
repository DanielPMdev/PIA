## 🟢 Bloque 2 — Función con múltiples outputs y examples

Una función puede devolver **varios valores** — solo necesitas una lista de outputs.
El parámetro `examples` añade ejemplos clicables para probar la app rápidamente.

```python
import gradio as gr

def analizar_texto(texto):
    palabras   = len(texto.split())
    caracteres = len(texto)
    mayusculas = sum(1 for c in texto if c.isupper())
    return (
        f'Palabras: {palabras}',
        f'Caracteres: {caracteres}',
        f'Mayúsculas: {mayusculas}'
    )

demo = gr.Interface(
    fn      = analizar_texto,
    inputs  = gr.Textbox(lines=5, label='Escribe un texto'),
    outputs = [
        gr.Textbox(label='Palabras'),
        gr.Textbox(label='Caracteres'),
        gr.Textbox(label='Mayúsculas'),
    ],
    title   = 'Analizador de texto',
    examples = [
        ['El gato come pescado'],
        ['Hola Mundo desde Valencia'],
        ['La INTELIGENCIA ARTIFICIAL cambia el MUNDO'],
    ]
)

demo.launch(share=True)
```

## 🟢 Bloque 3 — Slider y Number — modo live

`gr.Slider` permite seleccionar un valor numérico con una barra deslizante.
Con `live=True` la función se ejecuta **en tiempo real** sin necesidad de pulsar Submit.

```python
import gradio as gr

def calcular_imc(peso, altura):
    imc = peso / (altura ** 2)
    if imc < 18.5:   categoria = '🔵 Bajo peso'
    elif imc < 25:   categoria = '🟢 Normal'
    elif imc < 30:   categoria = '🟡 Sobrepeso'
    else:            categoria = '🔴 Obesidad'
    return round(imc, 2), categoria

demo = gr.Interface(
    fn      = calcular_imc,
    inputs  = [
        gr.Slider(40, 150, value=70, label='Peso (kg)'),
        gr.Slider(1.40, 2.20, value=1.75, step=0.01, label='Altura (m)'),
    ],
    outputs = [
        gr.Number(label='IMC'),
        gr.Textbox(label='Categoría'),
    ],
    live  = True,   # actualiza sin pulsar Submit
    title = 'Calculadora de IMC',
)

demo.launch(share=True)
```

## 🟢 Bloque 4 — Image — filtro escala de grises

`gr.Image` permite subir imágenes. Gradio las convierte automáticamente en arrays NumPy `[H, W, 3]`.

```python
import gradio as gr
import numpy as np

def escala_de_grises(imagen):
    # 1. Calculamos el shape para mostrarlo
    info_shape = f'Shape de la imagen: {imagen.shape}'

    # 2. Procesamos la imagen
    gris = np.mean(imagen, axis=2, keepdims=True)
    imagen_gris = np.repeat(gris, 3, axis=2).astype(np.uint8)

    # 3. DEVOLVEMOS LAS DOS COSAS
    return imagen_gris, info_shape

demo = gr.Interface(
    fn      = escala_de_grises,
    inputs  = gr.Image(label='Imagen original'),
    # DEFINIMOS DOS SALIDAS: una imagen y un texto
    outputs = [
        gr.Image(label='En escala de grises'),
        gr.Textbox(label='Información técnica (Shape)')
    ],
    title   = 'Filtro: escala de grises',
)

demo.launch()
```

**🔥 Reto:** Modifica la función para aplicar un tinte rojo:
- Canal R (índice 0): multiplica por 1.5
- Canal G (índice 1): multiplica por 0.5  
- Canal B (índice 2): multiplica por 0.5
- Usa `np.clip(resultado, 0, 255).astype(np.uint8)` para evitar desbordamiento

```python
import gradio as gr
import numpy as np

def tinte_rojo(imagen):
    # Convertimos la imagen a float para poder operar con decimales sin perder datos
    img_modificada = imagen.astype(float)

    # imagen llega como array numpy [H, W, 3] en formato RGB
    img_modificada[:, :, 0] *= 1.5  # Multiplica el canal Rojo (R)
    img_modificada[:, :, 1] *= 0.5  # Multiplica el canal Verde (G)
    img_modificada[:, :, 2] *= 0.5  # Multiplica el canal Azul (B)

    # Aseguramos que los valores se mantengan entre 0 y 255
    img_modificada = np.clip(img_modificada, 0, 255)

    # Convertimos de nuevo a formato de imagen (enteros de 8 bits)
    return img_modificada.astype(np.uint8)

demo = gr.Interface(
    fn      = tinte_rojo,
    inputs  = gr.Image(label='Imagen original'),
    outputs = gr.Image(label='Con tinte rojo'),
    title   = 'Filtro: Tinte Rojo',
)

demo.launch()
```

## 🟢 Bloque 5 — Audio — análisis de señal

`gr.Audio` devuelve una tupla `(sample_rate, datos)` donde datos es un array NumPy.

```python
import gradio as gr
import numpy as np

def info_audio(audio):
    if audio is None:
        return 'Sube un audio primero', '', ''
    sample_rate, datos = audio
    duracion   = len(datos) / sample_rate
    amplitud   = np.abs(datos).mean()
    return (
        f'{sample_rate} Hz',
        f'{duracion:.2f} segundos',
        f'{amplitud:.4f}',
    )

demo = gr.Interface(
    fn      = info_audio,
    inputs  = gr.Audio(label='Sube un audio o graba con el micrófono'),
    outputs = [
        gr.Textbox(label='Sample Rate'),
        gr.Textbox(label='Duración'),
        gr.Textbox(label='Amplitud media'),
    ],
    title = 'Analizador de audio',
)

demo.launch(share=True)
```

## 🟢 Bloque 6 — Layouts con gr.Blocks

`gr.Interface` es cómodo pero limitado. `gr.Blocks` te da control total:
- `gr.Row()` → elementos en horizontal
- `gr.Column()` → elementos en vertical
- `btn.click()` → conecta un botón a una función

Toda la lógica de eventos se define con `.click()`, `.change()`, `.submit()`...

```python
import gradio as gr

def mayusculas(texto): return texto.upper()
def minusculas(texto): return texto.lower()
def invertir(texto):   return texto[::-1]
def contar(texto):     return str(len(texto.split()))

with gr.Blocks(title='Transformador de texto') as demo:

    gr.Markdown('# ✏️ Transformador de texto')
    gr.Markdown('Escribe un texto y elige la transformación.')

    with gr.Column():
        entrada = gr.Textbox(label='Texto de entrada', lines=4, scale=2)
        salida  = gr.Textbox(label='Resultado',        lines=4, scale=2)

    with gr.Column():
        btn_may = gr.Button('MAYÚSCULAS', variant='primary')
        btn_min = gr.Button('minúsculas')
        btn_inv = gr.Button('Invertir')
        btn_cnt = gr.Button('Contar palabras')

    # Conectar cada botón a su función
    btn_may.click(fn=mayusculas, inputs=entrada, outputs=salida)
    btn_min.click(fn=minusculas, inputs=entrada, outputs=salida)
    btn_inv.click(fn=invertir,   inputs=entrada, outputs=salida)
    btn_cnt.click(fn=contar,     inputs=entrada, outputs=salida)

demo.launch(share=True)
```

Aquí es donde Gradio brilla de verdad: interfaz para cualquier modelo pre-entrenado.

> ⏳ **La primera ejecución descarga el modelo** (~260 MB). Puede tardar 1-2 minutos.
> Las siguientes veces va mucho más rápido porque Colab lo cachea.

```python
import gradio as gr
from transformers import pipeline

# Cargar el modelo UNA SOLA VEZ al arrancar (fuera de la función)
print('⏳ Cargando modelo...')
clasificador = pipeline('sentiment-analysis')
print('✅ Modelo listo')

def analizar_sentimiento(texto):
    if not texto.strip():
        return 'Escribe algo primero'
    resultado  = clasificador(texto)[0]
    etiqueta   = resultado['label']
    confianza  = resultado['score']
    emoji = '😊' if etiqueta == 'POSITIVE' else '😞'
    return f'{emoji} {etiqueta}  (confianza: {confianza:.1%})'

demo = gr.Interface(
    fn          = analizar_sentimiento,
    inputs      = gr.Textbox(
        lines=3,
        label='Escribe una frase en inglés',
        placeholder='I love this class!'
    ),
    outputs     = gr.Textbox(label='Sentimiento detectado'),
    title       = '😊 Analizador de sentimiento',
    description = 'Modelo: distilbert-base-uncased-finetuned-sst-2-english',
    examples    = [
        ['I love machine learning, it is amazing!'],
        ['This is terrible, I hate it.'],
        ['The weather is okay today.'],
        ['Valencia is a beautiful city!'],
    ]
)

demo.launch(share=True)
```

## 🤗 Bloque 9 — HuggingFace: Traducción ES → EN

El modelo MarianMT de Helsinki-NLP traduce entre decenas de idiomas.

```python
import gradio as gr
from transformers import MarianMTModel, MarianTokenizer
import torch

print('⏳ Cargando motor de traducción (Modo Manual Directo)...')

model_name = "Helsinki-NLP/opus-mt-es-en"

# Cargamos modelo y tokenizador sin usar 'pipeline'
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Si tienes GPU en Colab, lo movemos a la GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

print(f'✅ Cargado en {device.upper()}')

def traducir(texto_es):
    if not texto_es or not texto_es.strip():
        return 'Por favor, escribe algo...'

    try:
        # 1. Tokenizar el texto de entrada
        inputs = tokenizer(texto_es, return_tensors="pt", padding=True).to(device)

        # 2. Generar la traducción (los tokens de salida)
        translated_tokens = model.generate(**inputs)

        # 3. Convertir esos tokens en texto legible
        resultado = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)

        return resultado
    except Exception as e:
        return f"Error técnico: {str(e)}"

# Interfaz de Gradio
demo = gr.Interface(
    fn      = traducir,
    inputs  = gr.Textbox(lines=4, label='Texto en español', placeholder='Escribe aquí...'),
    outputs = gr.Textbox(lines=4, label='Traducción al inglés'),
    title   = '🌍 Traductor ES → EN (Motor Directo)',
    description = "Este traductor usa el modelo MarianMT directamente sin pasar por el pipeline de tareas."
)

demo.launch(share=True)
```

**🔥 Reto — Traductor con selector de idioma:**

Modifica el código para añadir un `gr.Dropdown` con opciones `['ES→EN', 'EN→ES', 'ES→FR']`
y carga el modelo correcto según la selección:
- ES→EN: `Helsinki-NLP/opus-mt-es-en`
- EN→ES: `Helsinki-NLP/opus-mt-en-es`  
- ES→FR: `Helsinki-NLP/opus-mt-es-ROMANCE`

```python
import gradio as gr
from transformers import pipeline

# Diccionario con los nombres de los modelos en Hugging Face
modelos = {
    'ES→EN': 'Helsinki-NLP/opus-mt-es-en',
    'EN→ES': 'Helsinki-NLP/opus-mt-en-es',
    'ES→FR': 'Helsinki-NLP/opus-mt-es-ROMANCE'
}

# Caché para almacenar los pipelines una vez cargados y ahorrar memoria
pipelines_cargados = {}

def traducir(direccion, texto):
    if not texto.strip():
        return ''

    # Cargar el modelo bajo demanda si no está en la caché
    if direccion not in pipelines_cargados:
        print(f"Cargando modelo para {direccion}...")
        pipelines_cargados[direccion] = pipeline('translation', model=modelos[direccion])

    traductor = pipelines_cargados[direccion]

    # El modelo es-ROMANCE requiere un token que indique el idioma de destino
    texto_procesado = texto
    if direccion == 'ES→FR':
        texto_procesado = f">>fr<< {texto}"

    resultado = traductor(texto_procesado, max_length=512)
    return resultado[0]['translation_text']

demo = gr.Interface(
    fn = traducir,
    inputs = [
        gr.Dropdown(
            choices=['ES→EN', 'EN→ES', 'ES→FR'],
            value='ES→EN', # Valor por defecto
            label='Dirección de traducción'
        ),
        gr.Textbox(lines=4, label='Texto de origen')
    ],
    outputs = gr.Textbox(lines=4, label='Traducción'),
    title = 'Traductor Multilingüe',
    examples = [
        ['ES→EN', 'El gato come pescado'],
        ['EN→ES', 'Good morning, how are you?'],
        ['ES→FR', 'La inteligencia artificial cambia el mundo'],
    ]
)

if __name__ == "__main__":
    demo.launch()
```

## 🤗 Bloque 10 — HuggingFace: NER — Reconocimiento de entidades

NER detecta entidades nombradas en el texto: personas (PER), lugares (LOC), organizaciones (ORG)...

```python
import gradio as gr
from transformers import pipeline

print('⏳ Cargando modelo NER...')
ner = pipeline('ner', aggregation_strategy="simple")
print('✅ Modelo listo')

def extraer_entidades(texto):
    if not texto.strip():
        return '', []
    entidades = ner(texto)
    # Tabla de resultados
    tabla = [
        [e['word'], e['entity_group'], f"{e['score']:.1%}"]
        for e in entidades
    ]
    # Texto con entidades marcadas
    texto_marcado = texto
    for e in entidades:
        texto_marcado = texto_marcado.replace(
            e['word'],
            f"**[{e['word']}]** ({e['entity_group']})"
        )
    return texto_marcado, tabla

with gr.Blocks(title='NER — Entidades nombradas') as demo:
    gr.Markdown('# 🔍 Reconocimiento de entidades nombradas')

    entrada = gr.Textbox(
        lines=3, label='Texto en inglés',
        placeholder='My name is Chelo and I work at IES Abastos in Valencia, Spain'
    )

    with gr.Row():
        texto_marcado = gr.Markdown(label='Texto con entidades marcadas')
        tabla = gr.Dataframe(
            headers=['Entidad', 'Tipo', 'Confianza'],
            label='Tabla de entidades'
        )

    gr.Button('🔍 Analizar', variant='primary').click(
        fn=extraer_entidades,
        inputs=entrada,
        outputs=[texto_marcado, tabla]
    )

    gr.Examples(
        examples=[
            ['My name is Chelo and I work at IES Abastos in Valencia, Spain'],
            ['Apple was founded by Steve Jobs in California'],
            ['The Eiffel Tower is in Paris, France'],
        ],
        inputs=entrada
    )

demo.launch(share=True)
```

## 🚀 Bloque 11 — Proyecto final

Elige **UNA** opción y construye la app. Al final comparte el enlace con la clase.

### Opción A — App de preguntas y respuestas
El usuario pega un contexto y hace una pregunta. La app responde.

### Opción B — Clasificador zero-shot personalizado  
El usuario escribe el texto Y las categorías. La app clasifica sin entrenamiento previo.

### Opción C — Tu propio modelo  
Conecta cualquier modelo que hayas entrenado en clase.

```python
# ── OPCIÓN A — Preguntas y Respuestas ──────────────────────────────────
import gradio as gr
from transformers import pipeline

print('⏳ Cargando modelo Q&A...')
qa = pipeline('question-answering')
print('✅ Modelo listo')

def responder(contexto, pregunta):
    if not contexto.strip() or not pregunta.strip():
        return 'Rellena el contexto y la pregunta', ''
    resultado = qa(question=pregunta, context=contexto)
    return resultado['answer'], f"{resultado['score']:.1%}"

with gr.Blocks(title='Q&A — Preguntas y respuestas') as demo:
    gr.Markdown('# ❓ Pregunta sobre cualquier texto')

    contexto = gr.Textbox(
        lines=6, label='Contexto (pega aquí el texto)',
        placeholder='Pega aquí el texto sobre el que quieres hacer preguntas...'
    )
    pregunta = gr.Textbox(
        label='Tu pregunta',
        placeholder='¿Cuál es...?'
    )

    with gr.Row():
        respuesta  = gr.Textbox(label='Respuesta')
        confianza  = gr.Textbox(label='Confianza')

    gr.Button('🔍 Responder', variant='primary').click(
        fn=responder,
        inputs=[contexto, pregunta],
        outputs=[respuesta, confianza]
    )

    gr.Examples(
        examples=[[
            'The Transformer architecture was introduced in 2017 by Vaswani et al. at Google Brain. It uses self-attention mechanisms and has become the foundation of models like BERT and GPT.',
            'Who created the Transformer?'
        ]],
        inputs=[contexto, pregunta]
    )

demo.launch(share=True)
```

```python
# ── OPCIÓN B — Clasificador zero-shot personalizado ────────────────────
import gradio as gr
from transformers import pipeline

print('⏳ Cargando modelo zero-shot...')
clasificador = pipeline('zero-shot-classification')
print('✅ Modelo listo')

def clasificar(texto, categorias_str):
    if not texto.strip() or not categorias_str.strip():
        return []
    categorias = [c.strip() for c in categorias_str.split(',')]
    resultado  = clasificador(texto, candidate_labels=categorias)
    # Devolver lista ordenada de [categoría, score]
    return [[label, f"{score:.1%}"]
            for label, score in zip(resultado['labels'], resultado['scores'])]

with gr.Blocks(title='Clasificador zero-shot') as demo:
    gr.Markdown('# 🏷️ Clasificador zero-shot personalizado')
    gr.Markdown('Define tus propias categorías separadas por comas.')

    with gr.Row():
        texto      = gr.Textbox(lines=4, label='Texto a clasificar')
        categorias = gr.Textbox(
            lines=2, label='Categorías (separadas por comas)',
            value='deportes, política, tecnología, cultura, economía'
        )

    resultado = gr.Dataframe(
        headers=['Categoría', 'Probabilidad'],
        label='Resultados (ordenados de mayor a menor probabilidad)'
    )

    gr.Button('🏷️ Clasificar', variant='primary').click(
        fn=clasificar,
        inputs=[texto, categorias],
        outputs=resultado
    )

    gr.Examples(
        examples=[
            ['El Real Madrid ganó la Champions League', 'deportes, política, tecnología, cultura'],
            ['El gobierno aprueba nuevos presupuestos', 'deportes, política, economía, salud'],
            ['La IA supera a humanos en diagnóstico médico', 'tecnología, salud, deportes, arte'],
        ],
        inputs=[texto, categorias]
    )

demo.launch(share=True)
```
